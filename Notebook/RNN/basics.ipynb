```python
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Input
```

```python
model = Sequential([
    Input(shape=(4,5)),
    SimpleRNN(3, activation='relu'),
    Dense(1, activation='sigmoid')
])
```

```python
model.summary()
```

```python
print('shape:-> ',model.get_weights()[0].shape)
print(model.get_weights()[0])
```

```python
print('shape:-> ',model.get_weights()[1].shape)
print(model.get_weights()[1])
```

```python
print('shape:-> ',model.get_weights()[2].shape)
print(model.get_weights()[2])
# prints the bias of the RNN layer
```

```python
print('shape:-> ',model.get_weights()[3].shape)
print(model.get_weights()[3])
```

```python
print('shape:-> ',model.get_weights()[4].shape)
print(model.get_weights()[4])
# prints the bias for the Dense layer
```

```python
print('shape:-> ',model.get_weights()[5].shape)
print(model.get_weights()[5])
# invalid index as there are only 5 sets of weights in the model
```
