```python
import numpy as np
```

```python
docs = ['You are what your thoughts have made you, so take care of what you think',
        'loving her was allways a loosing game, I knew but I still loved her',
        'either you are a scientist or believe in god can never be both',
        'I am slowing dying, I know but i can not do anything about it',
        'math and computer science are interrelated',
        'We have underestimated our brain for ever',
        'potan pahrera will be 3 division champion',
        'gravity is a discovery not a invention',
        'fossils band best singer ever',
        'transformer architechture',
		'kabib is best ufc fighter',
		'I dont think god exists',
		'trust is a bitch',
        'chemistry sucks']
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer(oov_token='<nothing>') # every word thats not in vocabhulary will be replaced ny <nothing>
```

```python
tokenizer.fit_on_texts(docs)
```

```python
tokenizer.word_index
# this shows the index of each word in the vocabhulary and the number of times it appears in the docs
# the most common word will have index 1 and so on
```

```python
tokenizer.word_counts
# frequency of each word in the docs
```

```python
tokenizer.document_count
# total no of sentences in the docs
```

```python
# generating sequences after tokenization
sequences = tokenizer.texts_to_sequences(docs)
sequences
```

```python
from tensorflow.keras.utils import pad_sequences
```

```python
sequences = pad_sequences(sequences,padding='post')
```
