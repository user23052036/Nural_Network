```python
import numpy as np
import pandas as pd
from sklearn.linear_model import Perceptron
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
or_data = pd.DataFrame()
and_data = pd.DataFrame()
xor_data = pd.DataFrame()
```

```python
or_data['input1'] = [0,0,1,1]
or_data['input2'] = [0,1,0,1]
or_data['output'] = [0,1,1,1]
```

```python
and_data['input1'] = [0,0,1,1]
and_data['input2'] = [0,1,0,1]
and_data['output'] = [0,0,0,1]
```

```python
xor_data['input1'] = [0,0,1,1]
xor_data['input2'] = [0,1,0,1]
xor_data['output'] = [0,1,1,0]
```

```python
or_perceptron = Perceptron()
and_perceptron = Perceptron()
xor_perceptron = Perceptron()
```

```python
X_or = or_data.iloc[:,0:2]
Y_or = or_data.iloc[:,-1]
```

```python
X_and = and_data.iloc[:,0:2]
Y_and = and_data.iloc[:,-1]
```

```python
X_xor = xor_data.iloc[:,0:2]
Y_xor = xor_data.iloc[:,-1]
```

```python
or_perceptron.fit(X_or,Y_or)
and_perceptron.fit(X_and,Y_and)
xor_perceptron.fit(X_xor,Y_xor)
```

## OR-graph

```python
or_perceptron.coef_
```

```python
or_perceptron.intercept_
```

```python
w = or_perceptron.coef_[0]
b = or_perceptron.intercept_[0]

x_vals = np.linspace(-1, 2, 100)

# derive y from w1*x1 + w2*x2 + b = 0
y_vals = -(w[0]*x_vals + b) / w[1]
```

```python
plt.figure(figsize=(4,3))
sns.scatterplot(
    x=or_data['input1'],
    y=or_data['input2'],
    hue=or_data['output'],
    s=200
)

plt.plot(x_vals, y_vals, color='black')

plt.xlim(-1, 2)
plt.ylim(-1, 2)
plt.grid(True)
plt.show()
```
