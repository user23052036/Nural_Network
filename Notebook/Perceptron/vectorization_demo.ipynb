```python
import time

a = np.random.rand(1000000)
b = np.random.rand(1000000)

tic = time.time()
c = np.dot(a,b)
toe = time.time()
print('vectrization: '+ str(round((toe-tic)*1000,3)) +'ms')

c=0
tic = time.time()
for i in range(1000000):
    c += a[i]*b[i]
toe = time.time()
print('for loop: '+ str(round((toe-tic)*1000,3)) +'ms')

# almost 200x slower for loop than dot product
```
